```python
# ==========================================
# 作業編號：HW09
# 作業名稱：Python 資料型別上機實作：兩場活動報名整理器
# 姓名：與那嶺藍
# 學號：D1430017
# 題目說明：整理「桌遊體驗」與「電影欣賞」兩場活動的原始報名文字，進行去重、集合比較與費用計算。
# ==========================================

# --- 測試一：使用原始資料 ---
print("【測試一】")
activity_names = ("桌遊體驗", "電影欣賞")
board_raw = " Amy, Ben, amy, Cora "
movie_raw = " ben, Dora, ERIC, ben "
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}

# 1. 讀取活動名稱，建立兩份整理後的串列（保留重複與原始順序）
act_board, act_movie = activity_names
board_list = board_raw.replace(" ", "").lower().split(",")
movie_list = movie_raw.replace(" ", "").lower().split(",")

# 2. 使用集合去重並比較名單
board_set = set(board_list)
movie_set = set(movie_list)

both_set = board_set & movie_set              # 共同參加者（交集）
all_set = board_set | movie_set               # 全部參加者（聯集）
only_board_set = board_set - movie_set        # 只參加桌遊者（差集）
only_movie_set = movie_set - board_set        # 只參加電影者（差集）

# 3. 建立排序串列，使用切片取得前兩位
board_sorted = sorted(board_set)
movie_sorted = sorted(movie_set)
all_sorted = sorted(all_set)
notify_top2 = all_sorted[:2]

# 4. 使用字典保存摘要，計算人數與應收費用
summary = {
    "桌遊人數": len(board_set),
    "電影人數": len(movie_set),
    "共同人數": len(both_set),
    "總不重複人數": len(all_set),
    "報名人次": len(board_set) + len(movie_set),
    "桌遊應收": len(board_set) * unit_fees[act_board],
    "電影應收": len(movie_set) * unit_fees[act_movie],
}
summary["總應收"] = summary["桌遊應收"] + summary["電影應收"]

# 5. 輸出完整摘要
print("HW09 活動報名摘要")
print(f"活動：{act_board}，每人 {unit_fees[act_board]:.2f} 元")
print(f"活動：{act_movie}，每人 {unit_fees[act_movie]:.2f} 元")
print(f"桌遊整理串列：{board_list}")
print(f"電影整理串列：{movie_list}")
print(f"桌遊去重名單：{board_sorted}")
print(f"電影去重名單：{movie_sorted}")
print(f"桌遊人數：{summary['桌遊人數']} 人")
print(f"電影人數：{summary['電影人數']} 人")
print(f"共同參加者：{sorted(both_set)}")
print(f"共同人數：{summary['共同人數']} 人")
print(f"只參加桌遊者：{sorted(only_board_set)}")
print(f"只參加電影者：{sorted(only_movie_set)}")
print(f"全部參加者：{all_sorted}")
print(f"總不重複人數：{summary['總不重複人數']} 人")
print(f"報名人次：{summary['報名人次']} 人次")
print(f"通知名單前兩位：{notify_top2}")
print(f"桌遊應收：{summary['桌遊應收']:.2f} 元")
print(f"電影應收：{summary['電影應收']:.2f} 元")
print(f"總應收：{summary['總應收']:.2f} 元")
print("-" * 40)


# --- 測試二：新增一位電影參加者 Frank ---
print("【測試二】")
activity_names = ("桌遊體驗", "電影欣賞")
board_raw = " Amy, Ben, amy, Cora "
movie_raw = " ben, Dora, ERIC, ben, Frank "
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}

act_board, act_movie = activity_names
board_list = board_raw.replace(" ", "").lower().split(",")
movie_list = movie_raw.replace(" ", "").lower().split(",")

board_set = set(board_list)
movie_set = set(movie_list)

both_set = board_set & movie_set
all_set = board_set | movie_set
only_board_set = board_set - movie_set
only_movie_set = movie_set - board_set

board_sorted = sorted(board_set)
movie_sorted = sorted(movie_set)
all_sorted = sorted(all_set)
notify_top2 = all_sorted[:2]

summary = {
    "桌遊人數": len(board_set),
    "電影人數": len(movie_set),
    "共同人數": len(both_set),
    "總不重複人數": len(all_set),
    "報名人次": len(board_set) + len(movie_set),
    "桌遊應收": len(board_set) * unit_fees[act_board],
    "電影應收": len(movie_set) * unit_fees[act_movie],
}
summary["總應收"] = summary["桌遊應收"] + summary["電影應收"]

print("HW09 活動報名摘要")
print(f"活動：{act_board}，每人 {unit_fees[act_board]:.2f} 元")
print(f"活動：{act_movie}，每人 {unit_fees[act_movie]:.2f} 元")
print(f"桌遊整理串列：{board_list}")
print(f"電影整理串列：{movie_list}")
print(f"桌遊去重名單：{board_sorted}")
print(f"電影去重名單：{movie_sorted}")
print(f"桌遊人數：{summary['桌遊人數']} 人")
print(f"電影人數：{summary['電影人數']} 人")
print(f"共同參加者：{sorted(both_set)}")
print(f"共同人數：{summary['共同人數']} 人")
print(f"只參加桌遊者：{sorted(only_board_set)}")
print(f"只參加電影者：{sorted(only_movie_set)}")
print(f"全部參加者：{all_sorted}")
print(f"總不重複人數：{summary['總不重複人數']} 人")
print(f"報名人次：{summary['報名人次']} 人次")
print(f"通知名單前兩位：{notify_top2}")
print(f"桌遊應收：{summary['桌遊應收']:.2f} 元")
print(f"電影應收：{summary['電影應收']:.2f} 元")
print(f"總應收：{summary['總應收']:.2f} 元")
print("-" * 40)


# --- 測試三：兩場去重後完全相同 ---
print("【測試三】")
activity_names = ("桌遊體驗", "電影欣賞")
board_raw = " Amy, Ben, amy, Cora "
movie_raw = " AMY, ben, CORA, amy "
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}

act_board, act_movie = activity_names
board_list = board_raw.replace(" ", "").lower().split(",")
movie_list = movie_raw.replace(" ", "").lower().split(",")

board_set = set(board_list)
movie_set = set(movie_list)

both_set = board_set & movie_set
all_set = board_set | movie_set
only_board_set = board_set - movie_set
only_movie_set = movie_set - board_set

board_sorted = sorted(board_set)
movie_sorted = sorted(movie_set)
all_sorted = sorted(all_set)
notify_top2 = all_sorted[:2]

summary = {
    "桌遊人數": len(board_set),
    "電影人數": len(movie_set),
    "共同人數": len(both_set),
    "總不重複人數": len(all_set),
    "報名人次": len(board_set) + len(movie_set),
    "桌遊應收": len(board_set) * unit_fees[act_board],
    "電影應收": len(movie_set) * unit_fees[act_movie],
}
summary["總應收"] = summary["桌遊應收"] + summary["電影應收"]

print("HW09 活動報名摘要")
print(f"活動：{act_board}，每人 {unit_fees[act_board]:.2f} 元")
print(f"活動：{act_movie}，每人 {unit_fees[act_movie]:.2f} 元")
print(f"桌遊整理串列：{board_list}")
print(f"電影整理串列：{movie_list}")
print(f"桌遊去重名單：{board_sorted}")
print(f"電影去重名單：{movie_sorted}")
print(f"桌遊人數：{summary['桌遊人數']} 人")
print(f"電影人數：{summary['電影人數']} 人")
print(f"共同參加者：{sorted(both_set)}")
print(f"共同人數：{summary['共同人數']} 人")
print(f"只參加桌遊者：{sorted(only_board_set)}")
print(f"只參加電影者：{sorted(only_movie_set)}")
print(f"全部參加者：{all_sorted}")
print(f"總不重複人數：{summary['總不重複人數']} 人")
print(f"報名人次：{summary['報名人次']} 人次")
print(f"通知名單前兩位：{notify_top2}")
print(f"桌遊應收：{summary['桌遊應收']:.2f} 元")
print(f"電影應收：{summary['電影應收']:.2f} 元")
print(f"總應收：{summary['總應收']:.2f} 元")
print("-" * 40)
```
